```python
import os
import glob
import json

# double check files are here
notebooks = glob.glob("*.ipynb")

if not notebooks:
    print("❌ Error: No .ipynb files found in this folder. Please check your path using %cd.")
else:
    print(f"Found {len(notebooks)} notebooks. Starting conversion...")

    for nb_file in notebooks:
        py_file = nb_file.replace(".ipynb", ".py")

        with open(nb_file, "r", encoding="utf-8") as f:
            try:
                nb_data = json.load(f)
            except Exception as e:
                print(f"Skipping broken file {nb_file}: {e}")
                continue

        with open(py_file, "w", encoding="utf-8") as out:
            out.write(f"# Converted from {nb_file}\n\n")
            for cell in nb_data.get("cells", []):
                if cell.get("cell_type") == "code":
                    for line in cell.get("source", []):
                        # Strip out terminal/magic commands (!pip, %matplotlib, etc.)
                        if not line.strip().startswith(('!', '%', 'get_ipython')):
                            out.write(line)
                    out.write("\n\n")

        print(f"✅ Converted: {nb_file} -> {py_file}")
```

```python
import os
import glob

output_file = "master_code_bundle.txt"

# This grabs all converted .py files in your current folder
files = glob.glob("*.py")

with open(output_file, "w", encoding="utf-8") as outfile:
    for f in files:
        # Avoid reading the output file itself if it accidentally ends with .py
        if f == "unified_output.py" or f == "unified_pipeline.py":
            continue

        outfile.write(f"\n\n# ==========================================\n")
        outfile.write(f"# SOURCE FILE: {f}\n")
        outfile.write(f"# ==========================================\n\n")

        with open(f, "r", encoding="utf-8") as infile:
            for line in infile:
                # Strip out Colab specific system commands that break standard Python
                if line.strip().startswith(('!', '%', 'get_ipython')):
                    continue
                outfile.write(line)

print(f"Successfully bundled {len(files)} Python files into '{output_file}'!")
```
