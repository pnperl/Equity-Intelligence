```python
import os
import glob
import json

# double check files are here
notebooks = glob.glob("*.ipynb")

if not notebooks:
    print("❌ Error: No .ipynb files found in this folder. Please check your path using %cd.")
else:
    print(f"Found {len(notebooks)} notebooks. Starting conversion...")

    for nb_file in notebooks:
        py_file = nb_file.replace(".ipynb", ".py")

        with open(nb_file, "r", encoding="utf-8") as f:
            try:
                nb_data = json.load(f)
            except Exception as e:
                print(f"Skipping broken file {nb_file}: {e}")
                continue

        with open(py_file, "w", encoding="utf-8") as out:
            out.write(f"# Converted from {nb_file}\n\n")
            for cell in nb_data.get("cells", []):
                if cell.get("cell_type") == "code":
                    for line in cell.get("source", []):
                        # Strip out terminal/magic commands (!pip, %matplotlib, etc.)
                        if not line.strip().startswith(('!', '%', 'get_ipython')):
                            out.write(line)
                    out.write("\n\n")

        print(f"✅ Converted: {nb_file} -> {py_file}")
```
